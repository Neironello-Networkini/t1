```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import easyocr

def detect_text_easyocr(image_path):
    reader = easyocr.Reader(['ru', 'en'])
    results = reader.readtext(image_path)

    image = cv2.imread(image_path)
    for (bbox, text, prob) in results:
        (top_left, top_right, bottom_right, bottom_left) = bbox
        top_left = tuple(map(int, top_left))
        bottom_right = tuple(map(int, bottom_right))

        cv2.rectangle(image, top_left, bottom_right, (0, 255, 0), 2)
        cv2.putText(image, text, (top_left[0], top_left[1] - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
    return image

def detect_text_opencv(image_path):
    image = cv2.imread(image_path)
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # фильтрация - бинаризация и морфология
    _, binary = cv2.threshold(gray, 150, 255, cv2.THRESH_BINARY_INV)
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
    morph = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

    # контур
    contours, _ = cv2.findContours(morph, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)

        if w > 30 and h > 10:
            cv2.rectangle(image, (x, y), (x+w, y+h), (255, 0, 0), 2)
    return image

def plot_images(img1, title1, img2, title2):
    plt.figure(figsize=(12,6))

    plt.subplot(1,2,1)
    plt.imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
    plt.title(title1)
    plt.axis('off')

    plt.subplot(1,2,2)
    plt.imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
    plt.title(title2)
    plt.axis('off')

    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    image_path = '../4_t_2.jpg'  # путь

    easyocr_detected = detect_text_easyocr(image_path)
    opencv_detected = detect_text_opencv(image_path)

    plot_images(easyocr_detected, "EasyOCR (CRAFT)", opencv_detected, "OpenCV (Contours)")
```

```python
import os
import json
import numpy as np
from PIL import Image

from paddleocr import PaddleOCR

# --- КОНФИГУРАЦИЯ ---
image_path = '../a_1_1.jpg'

output_image_dir = 'detected_text_regions'

output_json_path = 'detection_results.json'

use_gpu_config = False

def run_text_detection_and_save(img_path, out_img_dir, out_json_path, use_gpu=False):

    ocr_detector = PaddleOCR(use_angle_cls=True, lang='ru', use_gpu=use_gpu, rec=False, show_log=False)
    print("PaddleOCR инициализирован.")

    if not os.path.exists(img_path):
        print(f"Ошибка: Файл изображения не найден по пути: {img_path}")
        return

    print(f"Загрузка изображения: {img_path}")
    try:
        original_image = Image.open(img_path).convert('RGB')
    except Exception as e:
        print(f"ERROR: {e}")
        return

    dt_boxes = ocr_detector.ocr(img_path, cls=True, det=True, rec=False)
    print(f"Детект {len(dt_boxes)} .")

    os.makedirs(out_img_dir, exist_ok=True)
    print(f"ДИРР СОЗД: {out_img_dir}")

    output_data = []
    if dt_boxes and len(dt_boxes) > 0:

        detection_results = dt_boxes[0]

        for i, box_coords in enumerate(detection_results):
            box_id = i + 1

            box_points = np.array(box_coords).astype(np.int32)

            # (axis-aligned box)
            x_coords = box_points[:, 0]
            y_coords = box_points[:, 1]
            left, top = np.min(x_coords), np.min(y_coords)
            right, bottom = np.max(x_coords), np.max(y_coords)

            # запас обрезки
            padding = 2
            crop_box = (
                max(0, left - padding),
                max(0, top - padding),
                min(original_image.width, right + padding),
                min(original_image.height, bottom + padding)
            )

            try:
                cropped_image = original_image.crop(crop_box)
            except Exception as e:
                 print(f"error обрезка для бокса {box_id}: {e}")
                 continue

            png_filename = f"{box_id}.png"
            png_filepath = os.path.join(out_img_dir, png_filename)
            try:
                cropped_image.save(png_filepath, "PNG")
            except Exception as e:
                print(f"ошибка сейвинга {png_filepath}: {e}")
                continue

            record = {
                "id": box_id,
                "coordinates": box_points.tolist(),
                "filename": png_filename
            }
            output_data.append(record)
            # print(f"  Обработан {box_id}, сейв как {png_filename}")

    print(f"Сохранение результатов в JSON файл: {out_json_path}")
    try:
        with open(out_json_path, 'w', encoding='utf-8') as f:
            json.dump(output_data, f, ensure_ascii=False, indent=4)
        print("Готово!")
    except Exception as e:
        print(f"Ошибка сохранения JSON файла: {e}")

if __name__ == "__main__":
    run_text_detection_and_save(image_path, output_image_dir, output_json_path, use_gpu=use_gpu_config)
```
